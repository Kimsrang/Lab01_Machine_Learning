```python
import os

# Base folder inside data
project_base = "data/project"

project_dirs = [
    "src/core",
    "src/api",
    "src/data",
    "src/models",
    "src/utils",
    "src/config",
    "tests/unit",
    "tests/integration",
    "tests/e2e",
    "notebooks",
    "docs/api",
    "docs/usage",
    "docs/development",
    "conf",
    "scripts",
    ".github/workflows",
    "requirements",
]

project_files = [
    "Dockerfile",
    "docker-compose.yml",
    "Makefile",
    "pyproject.toml",
    "requirements/base.txt",
    "requirements/dev.txt",
    "requirements/prod.txt",
]

# Create directories
for d in project_dirs:
    os.makedirs(os.path.join(project_base, d), exist_ok=True)

# Create empty files
for f in project_files:
    file_path = os.path.join(project_base, f)
    os.makedirs(os.path.dirname(file_path), exist_ok=True)
    with open(file_path, "w") as fp:
        fp.write("")

print(f"✅ Project folder created at: {os.path.abspath(project_base)}")
```
