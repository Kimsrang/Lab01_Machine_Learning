```python
import os
base_dir = "data"

# Bronze layer: years 2020–2025, months 01–12
years = range(2020, 2026)
months = [f"{m:02d}" for m in range(1, 13)]

# Create Bronze folders
for year in years:
    for month in months:
        path = os.path.join(base_dir, "bronze", str(year), month)
        os.makedirs(path, exist_ok=True)
        print(f"Created: {path}")

# Create Silver and Gold folders
silver_path = os.path.join(base_dir, "silver")
gold_path = os.path.join(base_dir, "gold")

os.makedirs(silver_path, exist_ok=True)
os.makedirs(gold_path, exist_ok=True)
os.makedirs("notebooks", exist_ok=True)
# Create empty placeholder files

open(os.path.join(silver_path, "generation_forecast.db"), "a").close()
open(os.path.join(gold_path, "data_health.csv"), "a").close()
open(os.path.join(gold_path, "min_max_generation.csv"), "a").close()
open("notebooks/visualization.ipynb", "a").close()
print("\n Folder structure created successfully inside 'data/'")


```
